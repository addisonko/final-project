```python
import pandas as pd
from sqlalchemy import text
%run db_config.ipynb

CSV_URL = (
    "https://raw.githubusercontent.com/datasets/s-and-p-500-companies/"
    "master/data/constituents.csv"
)

def load_constituents() -> pd.DataFrame:
    df = pd.read_csv(CSV_URL)
    return df[["Symbol", "Security"]].dropna()

def seed_stocks(df: pd.DataFrame) -> None:
    engine = get_engine()
    inserted = 0
    skipped = 0

    with engine.begin() as conn:
        for _, row in df.iterrows():
            symbol = str(row["Symbol"]).strip()
            name = str(row["Security"]).strip()

            exists = conn.execute(
                text("SELECT 1 FROM stocks WHERE symbol = :symbol"),
                {"symbol": symbol},
            ).fetchone()

            if exists:
                skipped += 1
                continue

            conn.execute(
                text("INSERT INTO stocks (symbol, name) VALUES (:symbol, :name)"),
                {"symbol": symbol, "name": name},
            )
            inserted += 1
            print(f"INSERT {symbol}")

    print(f"\nDone. inserted={inserted}, skipped={skipped}")

if __name__ == "__main__":
    df = load_constituents()
    print(f"Loaded {len(df)} constituents")
    seed_stocks(df)
```
